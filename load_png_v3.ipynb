```python
import matplotlib.pyplot as plt
import numpy as np

idir = 'collocated_v3'
ifile = f'{idir}/S1A_EW_GRDM_1SDH_20250602T120218_20250602T120318_059467_0761D4_E361.SAFE#1_0.png'
img = plt.imread(ifile)

hh = img[:, :, 0]
hv = img[:, :, 2]
sic = np.clip(img[:, :, 1]*255 - 2, 0, 100)
mask = np.where(img[:, :, 1]*255 < 2, img[:, :, 1]*255, 2)
class_ = 255 - img[:, :, 3]*255

titles = ['HH', 'HV', 'SIC\n[0 - 100]', 'MASK\n0-bad, 1-land, 2-good', 'CLASS\n[0 - 5]']
arrays = [hh, hv, sic, mask, class_]
fig, axs = plt.subplots(1, 5, figsize=(12, 3))
for i in range(5):
    im = axs[i].imshow(arrays[i], interpolation='nearest')
    axs[i].set_title(titles[i])
    axs[i].axis('off')
    plt.colorbar(im, ax=axs[i], shrink=0.5)
plt.tight_layout()
```
